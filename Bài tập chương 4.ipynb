## TH 4.1

Viết lớp `HinhCau` có bán kính `ban_kinh`, các phương thức `dien_tich()` ($4\pi r^2$), `the_tich()` ($\frac43\pi r^3$) và `__str__`. Hàm tạo phải báo `ValueError` nếu bán kính không dương. Kiểm tra với $r=2$ (diện tích $16\pi\approx50{,}265$, thể tích $\frac{32\pi}{3}\approx33{,}510$) và $r=-1$.

```python
import math

class HinhCau:
    def __init__(self, ban_kinh: float):
        if ban_kinh <= 0:
            raise ValueError("Bán kính phải lớn hơn 0")
        self.ban_kinh = ban_kinh
        
    def dien_tich(self) -> float:
        return 4 * math.pi * self.ban_kinh ** 2

    def the_tich(self) -> float:
        return (4 / 3) * math.pi * self.ban_kinh ** 3

    def __str__(self) -> str:
        return f"Hình cầu có bán kính {self.ban_kinh}"

try:
    hc = HinhCau(2)

    print(f"Diện tích: {hc.dien_tich():.3f}")
    print(f"Thể tích: {hc.the_tich():.3f}")

except ValueError as e:
    print("Lỗi:", e)

try:
    hc = HinhCau(-1)

    print(f"Diện tích: {hc.dien_tich():.3f}")
    print(f"Thể tích: {hc.the_tich():.3f}")

except ValueError as e:
    print("Lỗi:", e)
```

## TH 4.3

Viết lớp `TaiKhoan` (chủ tài khoản, số dư ban đầu mặc định $0$ và *không âm*) với `nap(tien)`, `rut(tien)`, danh sách `lich_su` riêng từng tài khoản. Định nghĩa ngoại lệ `SoDuKhongDu` (kế thừa `Exception`) và dùng nó khi rút quá số dư; số tiền không dương thì báo `ValueError`. Kiểm tra bất biến ``số dư $\ge0$'' (Mệnh đề \ref{md:batbien}) với dãy: số dư đầu $1 000 000$, nạp $500 000$, rút $200 000$, rút $2 000 000$.

```python
class TaiKhoan:
    def __init__(self, chu_tai_khoan, so_du=0):
        if so_du < 0:
            raise ValueError("Số dư không được âm")

        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du
        self.lich_su = []

    def nap(self, tien):
        if tien <= 0:
            raise ValueError("Tiền nạp phải lớn hơn 0")

        self.so_du += tien
        self.lich_su.append(f"Nạp {tien}")

    def rut(self, tien):
        if tien <= 0:
            raise ValueError("Tiền rút phải lớn hơn 0")

        if tien > self.so_du:
            raise SoDuKhongDu("Số dư không đủ")

        self.so_du -= tien
        self.lich_su.append(f"Rút {tien}")

class SoDuKhongDu(Exception):
    pass

tk = TaiKhoan("Anh", 1000000)

try:
    tk.nap(500000)
    tk.rut(200000)
    tk.rut(2000000)

except (ValueError, SoDuKhongDu) as e:
    print("Lỗi:", e)

assert tk.so_du >= 0

print("Chủ tài khoản:", tk.chu_tai_khoan)
print("Số dư:", tk.so_du)
print("Lịch sử:", tk.lich_su)
```

## TH 4.5

Thiết kế lớp cha `HinhHoc` có phương thức `dien_tich()` chỉ báo `NotImplementedError` (buộc lớp con phải cài đặt) và `__str__`; ba lớp con `HinhTron(r)`, `HinhChuNhat(a, b)`, `HinhTamGiac(a, b, c)` (công thức Heron). Với danh sách gồm hình tròn $r=1$, chữ nhật $3\times4$ và tam giác $(3,4,5)$: in các hình theo diện tích tăng dần và tổng diện tích (phải bằng $\pi+18$), nhờ *đa hình*.

```python
class HinhHoc:
    def dien_tich(self):
        raise ValueError("chưa cài đặt dien_tich()")

    def __str__(self):
        return f"{self.__class__.__name__}: diện tích = {self.dien_tich():.3f}"

class HinhTron(HinhHoc):
    def __init__(self, r):
        self.r = r

    def dien_tich(self):
        return math.pi * self.r ** 2

class HinhChuNhat(HinhHoc):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def dien_tich(self):
        return self.a * self.b

class HinhTamGiac(HinhHoc):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def dien_tich(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p-self.a) * (p-self.b) * (p-self.c))

ds = [
    HinhTron(1),
    HinhChuNhat(3, 4),
    HinhTamGiac(3, 4, 5)
]

ds.sort(key=lambda h: h.dien_tich())

for h in ds:
    print(h)

tong = sum(h.dien_tich() for h in ds)

print(f"Tổng diện tích: {tong:.3f}")
print(f"Kiểm tra: pi + 18 = {math.pi + 18:.3f}")
```

## TH 4.7

Với tệp `data/goldprice.csv` (giá vàng theo tháng, cột `Date` dạng `yyyy-mm`, `Price`), tính giá trung bình của từng năm đủ $12$ tháng, rồi mức tăng (%) so với năm liền trước. In ba năm tăng mạnh nhất và năm giảm mạnh nhất. (Đáp số biết trước: năm $1980$ tăng mạnh nhất, năm $1981$ giảm mạnh nhất.)

```python
import csv
du_lieu = {}

with open("goldprice.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    for row in reader:
        nam, thang = row["Date"].split("-")
        nam = int(nam)
        thang = int(thang)
        gia = float(row["Price"])

        if nam not in du_lieu:
            du_lieu[nam] = {}

        du_lieu[nam][thang] = gia

gia_trung_binh = {}

for nam, cac_thang in du_lieu.items():
    if len(cac_thang) == 12:
        gia_trung_binh[nam] = sum(cac_thang.values()) / 12

ty_le = {}

for nam in gia_trung_binh:
    if nam - 1 in gia_trung_binh:
        nam_truoc = gia_trung_binh[nam - 1]
        nam_nay = gia_trung_binh[nam]

        ty_le[nam] = (nam_nay - nam_truoc) / nam_truoc * 100

top_3 = sorted(ty_le.items(),
               key=lambda x: x[1],
               reverse=True)[:3]

print("Ba năm tăng mạnh nhất:")
for nam, phan_tram in top_3:
    print(f"Năm {nam}: tăng {phan_tram:.2f}%")

nam_giam = min(ty_le, key=ty_le.get)

print("\nNăm giảm mạnh nhất:")
print(f"Năm {nam_giam}: {ty_le[nam_giam]:.2f}%")
```

```python
class DongkhongHopLe(Exception):
    pass
    
dong_hop_le = []
dong_loi = []

with open("luong.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    stt = 2

    for row in reader:
        try:
            nam_kn = int(row["nam_kn"])
            luong = float(row["luong"])

            dong_hop_le.append((nam_kn, luong))

        except (ValueError, TypeError):
            dong_loi.append(stt)

        stt += 1

print("Các dòng hợp lệ:")
for dong in dong_hop_le:
    print(dong)

print("Các dòng lỗi:", dong_loi)

if len(dong_hop_le) > 0:
    tong = 0

    for nam, luong in dong_hop_le:
        tong += luong

    trung_binh = tong / len(dong_hop_le)

    print("Lương trung bình:", trung_binh)
```
