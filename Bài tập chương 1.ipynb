## Bài 1.1 — Dự đoán kết quả

Với `x = 23` và `y = 6`, hãy tính tay giá trị các biểu thức `x / y`, `x // y`, `x % y`, `x ** 2`,
`-x // y`, `-x % y`, `2 ** 3 ** 2`, `(2 ** 3) ** 2`; sau đó kiểm tra lại bằng máy.





```python
x = 23
y = 6

print("x / y =", x / y)
print("x // y =", x // y)
print("x % y =", x % y)
print("x ** 2 =", x ** 2)
print("-x // y =", -x // y)
print("-x % y =", -x % y)
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)
```

## Bài 1.3 — Dự đoán kết quả

Dự đoán đầu ra của bốn dòng dưới đây rồi chạy để kiểm tra. Chú ý các giá trị làm tròn và phép ép kiểu; giải thích *vì sao* `round(2.5)` và `round(3.5)` không cho kết quả ``tăng một đơn vị'' như nhau.

### TH 1.1. Tính chu vi và diện tích tam giác

Cho tam giác có ba cạnh `a`, `b`, `c`.
- Tính chu vi tam giác.
- Tính diện tích tam giác bằng công thức Heron.
**Công thức:**
Chu vi: `P = a + b + c`
Nửa chu vi: `p = (a + b + c) / 2`
Diện tích: `S = √(p(p-a)(p-b)(p-c))`

```python

a, b, c = 13, 14, 15

chu_vi = a + b + c
p = chu_vi / 2
dien_tich = (p * (p-a) * (p-b) * (p-c)) ** 0.5

print("Chu vi:", chu_vi)
print("Diện tích:", dien_tich)

a, b, c = 3, 4, 5

chu_vi = a + b + c
p = chu_vi / 2
dien_tich = (p * (p-a) * (p-b) * (p-c)) ** 0.5

print("Chu vi:", chu_vi)
print("Diện tích:", dien_tich)
```


### TH 1.2. Tính lãi suất tiết kiệm

Gửi tiết kiệm kỳ hạn 1 tháng với số tiền ban đầu là
50 000 000 đồng, lãi suất 6%/năm.

Yêu cầu:
- Lãi suất mỗi tháng bằng 6%/12 của số dư.
- Tiền lãi mỗi tháng được cộng vào tiền gốc.
- In số dư sau 1, 2 và 3 tháng.
- Làm tròn đến đồng, có dấu ngăn cách hàng nghìn.
Số dư mới = Số dư cũ × (1 + 0.06/12)


```python

so_tien = 50_000_000
lai_suat = 0.06 / 12

so_tien *= (1 + lai_suat)
print(f"Sau 1 tháng: {so_tien:,.0f} đồng")

so_tien *= (1 + lai_suat)
print(f"Sau 2 tháng: {so_tien:,.0f} đồng")

so_tien *= (1 + lai_suat)
print(f"Sau 3 tháng: {so_tien:,.0f} đồng")
```


### TH 1.3. Đổi phút thành ngày, giờ, phút

Cho tổng số phút `tong_phut`.

Yêu cầu:
- Chuyển tổng số phút thành dạng:
  `x ngày y giờ z phút`.
**Quy đổi:**


### TH 1.4. Mô phỏng máy rút tiền ATM

Một máy ATM cần trả số tiền `so_tien = 1_880_000`
đồng bằng các tờ tiền có mệnh giá:

500 000, 200 000, 100 000, 50 000,
20 000, 10 000 đồng.

Yêu cầu:
- Ưu tiên sử dụng các mệnh giá lớn trước.
- In số tờ tiền của từng mệnh giá.
- Kiểm tra tổng giá trị các tờ tiền đã trả
  bằng số tiền ban đầu.

Sử dụng phép chia nguyên `//` và chia lấy dư `%`.


```python

so_tien = 1_880_000
con_lai = so_tien

to_500 = con_lai // 500_000
con_lai %= 500_000

to_200 = con_lai // 200_000
con_lai %= 200_000

to_100 = con_lai // 100_000
con_lai %= 100_000

to_50 = con_lai // 50_000
con_lai %= 50_000

to_20 = con_lai // 20_000
con_lai %= 20_000

to_10 = con_lai // 10_000
con_lai %= 10_000

print("500.000 đồng:", to_500, "tờ")
print("200.000 đồng:", to_200, "tờ")
print("100.000 đồng:", to_100, "tờ")
print("50.000 đồng:", to_50, "tờ")
print("20.000 đồng:", to_20, "tờ")
print("10.000 đồng:", to_10, "tờ")
print("Số tiền còn lại:", con_lai)

tong = (to_500 * 500_000
        + to_200 * 200_000
        + to_100 * 100_000
        + to_50 * 50_000
        + to_20 * 20_000
        + to_10 * 10_000)

print("Tổng tiền đã trả:", tong)
print("Kiểm tra:", tong == so_tien)
```


### TH 1.5. Tách chữ số và đảo ngược số nguyên

Cho số nguyên bốn chữ số `n = 7539`.

Yêu cầu:
- Chỉ sử dụng phép chia nguyên `//` và
  phép chia lấy dư `%` để tách chữ số.
- In chữ số hàng nghìn, trăm, chục, đơn vị.
- In từng chữ số.
- Tính tổng các chữ số.
- Đảo ngược số thành `9357`.
- Kiểm tra bằng `int(str(n)[::-1])`.


```python

n = 7539

nghin = n // 1000
tram = (n // 100) % 10
chuc = (n // 10) % 10
don_vi = n % 10

print("Hàng nghìn:", nghin)
print("Hàng trăm:", tram)
print("Hàng chục:", chuc)
print("Hàng đơn vị:", don_vi)

tong = nghin + tram + chuc + don_vi
print("Tổng các chữ số:", tong)

dao_nguoc = (don_vi * 1000
             + chuc * 100
             + tram * 10
             + nghin)

print("Số đảo ngược:", dao_nguoc)
print("Kiểm tra:", int(str(n)[::-1]))
```
