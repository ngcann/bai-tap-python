## TH 3.2

Điểm của $8$ giám khảo là `[7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]`. Để giảm ảnh hưởng của giá trị bất thường, bỏ đi $n=2$ điểm cao nhất và $n=2$ điểm thấp nhất rồi tính trung bình các điểm còn lại (*trung bình cắt đầu--cuối*). So sánh với trung bình thông thường; *không* sửa danh sách gốc.

```python
diem = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]
tb = sum(diem) / len(diem)
diem_sap_xep = sorted(diem)

diem_con_lai = diem_sap_xep[2:-2]
tb_sau = sum(diem_con_lai) / len(diem_con_lai)

print("Danh sách gốc:", diem)
print("Danh sách đã sắp xếp:", diem_sap_xep)
print("Điểm còn lại:", diem_con_lai)
print("Trung bình thông thường:", tb)
print("Trung bình sau:", tb_sau)
```

## TH 3.4

Số giờ ôn tập $x$ và điểm thi $y$ của $5$ sinh viên là `[(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]`. Tìm đường hồi quy tuyến tính $y=a+bx$ theo công thức $b=\dfrac{\sum(x_i-\bar x)(y_i-\bar y)}{\sum(x_i-\bar x)^2}$, $a=\bar y-b\bar x$; dự đoán điểm của sinh viên ôn $6$ giờ; kiểm tra bằng `statistics.linear_regression`.

```python
import statistics
x = [1, 2, 3, 4, 5]
y = [4.0, 5.5, 6.0, 7.5, 8.0]

x_tb = sum(x) / len(x)
y_tb = sum(y) / len(y)

tu_b = 0
mau_b = 0

for i in range(len(x)):
    tu_b += (x[i] - x_tb) * (y[i] - y_tb)
    mau_b += (x[i] - x_tb)**2

b = tu_b / mau_b
a = y_tb - b * x_tb

print(f"Phương trình: y = {a:.2f} + {b:.2f}x")
print(f"Điểm dự đoán 6 giờ ôn: {a + b*6:.2f}")

print(statistics.linear_regression(x, y))
```

## TH 3.6

Cho danh sách các cặp `(tên, điểm)`: `[("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]`. Quy đổi sang điểm chữ ($\ge8{,}5$: A; $\ge7$: B; $\ge5{,}5$: C; $\ge4$: D; còn lại F) rồi gom tên sinh viên theo điểm chữ bằng từ điển (dùng `defaultdict(list)`). In các nhóm theo thứ tự chữ cái.

```python
from collections import defaultdict

ds = [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6),
      ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]

nhom = defaultdict(list)

for ten, diem in ds:
    if diem >= 8.5:
        chu = "A"
    elif diem >= 7:
        chu = "B"
    elif diem >= 5.5:
        chu = "C"
    elif diem >= 4:
        chu = "D"
    else:
        chu = "F"

    nhom[chu].append(ten)

for chu in sorted(nhom):
    print(chu, ":", nhom[chu])
```

## TH 3.8

Doanh thu theo sản phẩm của hai cửa hàng: \texttt{{"ao": 120, "quan": 80, "non": 30}} và \texttt{{"ao": 90, "giay": 150, "non": 20}}. Hợp nhất thành tổng doanh thu theo sản phẩm, in theo thứ tự giảm dần; tìm các sản phẩm bán ở cả hai cửa hàng và các sản phẩm chỉ bán ở một cửa hàng (dùng phép toán tập hợp trên khoá).

```python
ch1 = {"ao": 120, "quan": 80, "non": 30}
ch2 = {"ao": 90, "giay": 150, "non": 20}

doanh_thu = ch1.copy()

for sp in ch2:
    if sp in doanh_thu :
        doanh_thu[sp] += ch2[sp]
    else:
        doanh_thu[sp] = ch2[sp]

ds_san_pham = sorted(doanh_thu, key=doanh_thu.get, reverse=True)

for sp in ds_san_pham:
    print(sp, ":", doanh_thu[sp])

s1 = set(ch1)
s2 = set(ch2)

print("Sản phẩm bán ở cả hai cửa hàng:", s1 & s2)
print("Sản phẩm chỉ bán ở một cửa hàng:", s1 ^ s2)
```

## TH 3.10 ∗

Dùng sàng Eratosthenes (cài đặt bằng `set`, theo Mệnh đề \ref{md:sang}) tìm các số nguyên tố không vượt quá $1000$: in số lượng, tổng, và số cặp *số nguyên tố sinh đôi* $(p,p+2)$ cùng $\le1000$. Kiểm tra: có $168$ số nguyên tố, tổng bằng $76 127$.

```python
n = 1000
snt = set(range(2, n ))
p = 2

while p * p <= n:
    if p in snt:
        for k in range(p * p, n , p):
            snt.discard(k)
    p += 1

print("Số lượng số nguyên tố:", len(snt))
print("Tổng các số nguyên tố:", sum(snt))

dem = 0
for p in snt:
    if p + 2 in snt:
        dem += 1

print("Số cặp số nguyên tố sinh đôi:", dem)
```
