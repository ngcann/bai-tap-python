## TH 2.3

Viết chương trình tính lương tuần như Ví dụ \ref{vd:luong} (lương $100 000$ đồng/giờ, giờ vượt $40$ tính gấp $1{,}5$) nhưng *nhập lại* cho đến khi người dùng gõ một số *không âm*: chữ hoặc số âm đều phải báo lỗi và hỏi lại. Kiểm tra với dãy gõ `abc`, `-2`, `45`.

```python
def luong(gio: float) -> float:
    if gio > 40:
        tien = 40 * 100000 + (gio - 40) * 100000 * 1.5
    else:
        tien = gio * 100000
    return tien


while True:
    try:
        gio = float(input("Nhập số giờ làm trong tuần: "))
        if gio < 0:
            print("Số giờ không được âm")
            continue
        break
    except ValueError:
        print("Nhập một số hợp lệ")

print(f"Lương tuần: {luong(gio)} đồng")
```

## TH 2.4

Nhập liên tục các số thực (bỏ qua và báo giá trị không phải số) cho đến khi gõ `xong`. In số giá trị hợp lệ, tổng, trung bình, giá trị lớn nhất và nhỏ nhất; nếu chưa nhập số nào thì báo ``chưa có dữ liệu''. Kiểm tra với dãy gõ: `7.5, 8, abc, 6.5, 9, xong`.

```python
def tinh_toan(ds: list) -> None:
    if len(ds) == 0:
        print("Chưa có dữ liệu")
    else:
        print("Số giá trị hợp lệ:", len(ds))
        print("Tổng:", sum(ds))
        print("Trung bình:", sum(ds) / len(ds))
        print("Giá trị lớn nhất:", max(ds))
        print("Giá trị nhỏ nhất:", min(ds))
        
ds = []

while True:
    x = input("Nhập số thực (gõ 'xong' để kết thúc): ")
    
    if x == "xong":
        break

    try:
        so = float(x)
        ds.append(so)
    except ValueError:
        print("Giá trị không phải số")

tinh_toan(ds)
```

## TH 2.7

Dãy Collatz: từ $n$, nếu $n$ chẵn thì $n\leftarrow n/2$, nếu lẻ thì $n\leftarrow3n+1$, lặp đến khi $n=1$. Với $n=27$, đếm số bước và tìm giá trị lớn nhất mà dãy đạt được (đáp án biết trước: $111$ bước, giá trị lớn nhất $9232$).

```python
def collatz(n: int) -> None:
    buoc = 0
    gia_tri_max = n

    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1

        buoc += 1

        if n > gia_tri_max:
            gia_tri_max = n

    return buoc, gia_tri_max


n = 27
buoc, gia_tri_max = collatz(n)

print("Số bước:", buoc)
print("Giá trị lớn nhất:",gia_tri_max)
```

## TH 2.10 ∗

Mở rộng Ví dụ \ref{vd:montepi}: ước lượng $\pi$ với $N=10^2,\ 10^4,\ 10^6$ điểm (dùng `random.seed(2026)` một lần ở đầu). In ước lượng và sai số tuyệt đối; kiểm tra bằng `assert` rằng sai số nhỏ hơn $6/\sqrt N$. Nhận xét: gấp $100$ lần số điểm thì sai số giảm khoảng mấy lần?

```python
import random
import math

def uoc_luong(N: int) -> float:
    M = 0

    for i in range(N):
        x = random.random()
        y = random.random()

        if x**2 + y**2 <= 1:
            M += 1

    return 4 * M / N


random.seed(2026)

for N in [10**2, 10**4, 10**6]:
    pi_uoc_luong = uoc_luong(N)
    sai_so = abs(pi_uoc_luong - math.pi)

    print(f"N = {N}")
    print(f"Ước lượng π = {pi_uoc_luong}")
    print(f"Sai số tuyệt đối = {sai_so}")
    print(f"Ngưỡng 6/N = {6/N}")

    assert sai_so < 6 / math.sqrt(N)
    print("Kiểm tra: Đạt")
    print()
```

Phương pháp Monte Carlo có sai số điển hình tỉ lệ nghịch với căn bậc hai của số điểm lấy mẫu. Do đó, khi tăng số điểm lên 100 lần, sai số có xu hướng giảm khoảng 10 lần.

## TH 2.11 ∗

Phương trình $f(x)=x^3-x-2=0$ có đúng một nghiệm trong $[1,2]$ vì $f(1)=-2<0<4=f(2)$ và $f$ tăng trên đoạn này. Cài đặt *phương pháp chia đôi*: giữ đoạn $[a,b]$ mà $f(a)f(b)<0$, thay bằng nửa đoạn chứa nghiệm cho tới khi $b-a<10^{-10}$. In nghiệm, số bước, và so sánh với kết quả lý thuyết: số bước cần thiết là $\lceil\log_2(1/10^{-10})\rceil$.

```python
def f(x: float) -> float:
    return x**3 - x - 2

def chia_doi(a: float, b: float, sai_so: float) -> None:
    buoc = 0
    while b - a >= sai_so:
        c = (a + b) / 2

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

        buoc += 1

    nghiem = (a + b) / 2
    ly_thuyet = math.log2(1 / sai_so)

    print("Nghiệm gần đúng:", nghiem)
    print("Số bước thực tế:", buoc)
    print("Số bước lý thuyết:", ly_thuyet)


chia_doi(1, 2, 10**-10)
```
